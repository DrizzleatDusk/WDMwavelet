```python
import numpy as np
from wdmwavelet.wdmwavelet import _rwdm, PI, SQRT2, _gen_rolloff
```

```python
nf = 6
nt = 5
n = nf * nt
steepness = 2

phi_tilde0 = [1, 1, .97, .243] + [0,] * 12
phi_tilde0 = np.array(phi_tilde0) * 2
phi_tilde1 = [0, 0, .243, .97, 1, 1, 1, .97, .243] + [0,] * 7
phi_tilde1 = np.array(phi_tilde1)
phi_tilde2 = [0,] * 7 + [.243, .97, 1, 1, 1, .97, .243, 0, 0]
phi_tilde2 = np.array(phi_tilde2)
phi_tilde3 = [0,]*12 + [ .243, .97, 1, 1]
phi_tilde3 = np.array(phi_tilde3) * 2
phi_tilde = np.vstack((phi_tilde0, phi_tilde1,
                       phi_tilde2, phi_tilde3)) / nt**.5 

x = np.random.randn(n)
x_tilde = np.fft.rfft(x, norm="ortho")
w = _rwdm(x, nt, nf, 2., steepness)
```

```python
x_tilde_my = np.zeros(16, dtype=complex)
for k in range(16):
    for m in range(4):
        for n in range(10):

            if m == 0 or m == 3:
                A = 1
            else: A = SQRT2

            if (m + n) % 2 == 0:
                A_mn = A
            else: A_mn = -1j * A

            awe = A_mn * w[m][n] * np.exp(-1j*PI * n * k / nt)

            x_tilde_my[k] += phi_tilde[m][k] * awe

x_tilde_my /= 2.
```
